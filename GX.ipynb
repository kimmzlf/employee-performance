```python
from great_expectations.data_context import FileDataContext
```

```python
context = FileDataContext.create(project_root_dir='./')
```

```python

datasource_name = 'csv_data_employee'
datasource = context.sources.add_pandas(datasource_name)


asset_name = 'data_employee'
path_to_data = 'data_clean.csv'
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer=path_to_data)


batch_request = asset.build_batch_request()
```

```python

expectation_suite_name = 'expectation_data_employee'
context.add_or_update_expectation_suite(expectation_suite_name)

validator = context.get_validator(
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name
)

validator.head()
```

## 1. to be unique

```python
# Memastikan kolom eployee_id bersifat unique
validator.expect_column_values_to_be_unique('employee_id')
```

## 2. to be between min_value and max_value

```python
# Memastikan colom performance_score berada dalam range tertentu
validator.expect_column_values_to_be_between(
    column='performance_score', min_value=1, max_value=5
)
```

## 3. to be in set

```python
# Memastikan kolom job_title memiliki value yang ditentukan
validator.expect_column_values_to_be_in_set('job_title', 
                                            ["Analyst", "Consultant", "Developer", "Engineer", "Manager", "Specialist", "Technician"])
```

## 4. to be in type list

```python
# Memastikan kolom age memiliki tipe data yang ditentukan
validator.expect_column_values_to_be_in_type_list('age', ['integer', 'int', 'int32', 'int64'])
```

## 5. to be greater than

```python
# Memastikan kolom age tidak lebih besar dari years_at_company
validator.expect_column_pair_values_a_to_be_greater_than_b(column_A='age', column_B='years_at_company')
```

## 6. to be in set

```python
# Memastikan value kolom resigned kebanyakan berada dalam nilai yang ditentukan
validator.expect_column_values_to_be_in_set(
    column='resigned', 
    value_set=[False], 
    mostly=0.80
)
```

## 7. row to be between

```python
# Memastikan jumlah kolom berada dalam range yang ditentukan 
validator.expect_table_row_count_to_be_between(min_value=0, max_value=300000)
```
